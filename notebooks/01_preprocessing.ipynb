## 5. Handling Duplicate Records

Duplicate records were checked using complete-row comparison.

No duplicate records were found in the raw dataset. The duplicate-removal operation is still applied to ensure the preprocessing pipeline remains robust and reproducible.

```python
# Remove duplicate records
df = df.drop_duplicates()

print("Dataset shape after duplicate removal:", df.shape)
```

## 6. Handling Missing BMI Values

The BMI feature contains missing observations. Instead of removing these records, missing BMI values are imputed using the median BMI calculated from the available observations.

Median imputation was selected because it is less sensitive to extreme values than mean imputation and helps preserve the number of observations available for subsequent analysis and model development.

```python
# Calculate the median BMI using available values
bmi_median = df["bmi"].median()

print("Median BMI:", bmi_median)
```

```python
df = df.drop_duplicates()
print("Dataset shape after duplicate removal:", df.shape)
```

```python
bmi_median = df["bmi"].median()
print("Median BMI:", bmi_median)
```

```python
# Fill missing BMI values using the median
df["bmi"] = df["bmi"].fillna(bmi_median)

print("Missing BMI values after imputation:", df["bmi"].isnull().sum())
```

```python
# Verify dataset shape after BMI imputation
print("Dataset shape after BMI imputation:", df.shape)
```

```python
# Verify missing values across the entire dataset
print("Missing values after preprocessing:")
print(df.isnull().sum())
```

## 7. Removing Unnecessary Columns

The `id` column is a unique identifier and does not represent a meaningful clinical or lifestyle feature for stroke prediction. Therefore, it is removed before further processing.

```python
# Remove the unique identifier
df = df.drop(columns=["id"])

print("Dataset shape after removing id:", df.shape)
print("\nRemaining columns:")
print(df.columns.tolist())
```

```python
# Inspect categorical columns and their unique values

categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].value_counts())
```

## 8. Encoding Categorical Variables

Categorical variables are converted into numerical representations so that they can be used by machine learning algorithms.

One-hot encoding is used for categorical features because the categories do not have an inherent numerical order. The target variable `stroke` is retained as a binary numerical variable.

```python
# Encode categorical variables using one-hot encoding

categorical_columns = df.select_dtypes(include="object").columns

df = pd.get_dummies(
    df,
    columns=categorical_columns,
    drop_first=False,
    dtype=int
)

print("Dataset shape after categorical encoding:", df.shape)
print("\nEncoded columns:")
print(df.columns.tolist())
```

```python
# Final dataset validation

print("Final dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDuplicate records:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
# Verify the target variable

print("Stroke target values:")
print(df["stroke"].value_counts())

print("\nTarget data type:", df["stroke"].dtype)
```

```python
# Save the processed dataset

output_path = "../dataset/processed/cleaned_stroke_dataset.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully.")
print("File:", output_path)
```
