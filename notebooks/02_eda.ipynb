# NeuroShield AI — Exploratory Data Analysis

### Stroke Risk Prediction Dataset

This notebook explores the stroke dataset using basic statistics and simple visualizations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import os

os.getcwd()
```

```python
df = pd.read_csv("../dataset/processed/cleaned_stroke_dataset.csv")
df.head()
```

```python
df.shape
```

```python
df.dtypes
```

## Summary

```python
df.describe()
```

## Stroke Distribution

```python
sns.countplot(x="stroke", data=df)
plt.title("Stroke Distribution")
plt.xlabel("Stroke")
plt.ylabel("Number of Patients")
plt.show()
```

## Observation
The dataset is highly imbalanced. Most people didn't have stroke, only a small number did.

## Age Distribution

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["age"], bins=20, kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.show()
```

# Age vs Stroke

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x="stroke", y="age", data=df)
plt.title("Age Distribution by Stroke Status")
plt.xlabel("Stroke")
plt.ylabel("Age")
plt.show()
```

## Average Glucose Level and Stroke

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x="stroke", y="avg_glucose_level", data=df)
plt.title("Average Glucose Level by Stroke Status")
plt.xlabel("Stroke")
plt.ylabel("Average Glucose Level")
plt.show()
```

### Observation

Patients with stroke generally have higher average glucose levels than patients without stroke.

## BMI and Stroke

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x="stroke", y="bmi", data=df)
plt.title("BMI Distribution by Stroke Status")
plt.xlabel("Stroke")
plt.ylabel("BMI")
plt.show()
```

BMI values fairly similar for patients with and without stroke

## Hypertension and Stroke

```python
sns.countplot(x="hypertension", hue="stroke", data=df)
plt.title("Hypertension and Stroke")
plt.xlabel("Hypertension")
plt.ylabel("Number of Patients")
plt.legend(title="Stroke")
plt.show()
```

# Observation
Most patients didn't have stroke both with and without hypertension. Number of stroke cases is much smaller.

```python
hypertension_stroke_rate = df.groupby("hypertension")["stroke"].mean() * 100
hypertension_stroke_rate
```

```python
hypertension_stroke_rate.plot(kind="bar")

plt.title("Stroke Rate by Hypertension")
plt.xlabel("Hypertension")
plt.ylabel("Stroke Rate (%)")
plt.xticks([0, 1], ["No", "Yes"], rotation=0)
plt.show()
```

## Stroke Rate by Heart Disease

```python
heart_disease_stroke_rate = df.groupby("heart_disease")["stroke"].mean() * 100
heart_disease_stroke_rate
```

stroke rate is higher among patients with heart disease.

```python
heart_disease_stroke_rate.plot(kind="bar")
plt.title("Stroke Rate by Heart Disease")
plt.xlabel("Heart Disease")
plt.ylabel("Stroke Rate (%)")
plt.xticks([0, 1], ["No", "Yes"], rotation=0)
plt.show()
```

# Observation
Hypertension → stroke rate: 3.97% vs 13.25%


Heart disease → stroke rate: 4.18% vs 17.03%

## Stroke Rate by Gender

```python
gender_stroke_rate = pd.DataFrame({
    "gender": ["Female", "Male", "Other"],
    "stroke_rate": [
        df.loc[df["gender_Female"] == 1, "stroke"].mean() * 100,
        df.loc[df["gender_Male"] == 1, "stroke"].mean() * 100,
        df.loc[df["gender_Other"] == 1, "stroke"].mean() * 100
    ]
})
gender_stroke_rate
```

## Stroke Rate by Smoking Status

```python
smoking_stroke_rate = pd.DataFrame({
    "smoking_status": ["Unknown", "Formerly Smoked", "Never Smoked", "Smokes"],
    "stroke_rate": [
        df.loc[df["smoking_status_Unknown"] == 1, "stroke"].mean() * 100,
        df.loc[df["smoking_status_formerly smoked"] == 1, "stroke"].mean() * 100,
        df.loc[df["smoking_status_never smoked"] == 1, "stroke"].mean() * 100,
        df.loc[df["smoking_status_smokes"] == 1, "stroke"].mean() * 100
    ]
})
smoking_stroke_rate
```

```python
smoking_stroke_rate.plot(
    x="smoking_status",
    y="stroke_rate",
    kind="bar",
    legend=False
)
plt.title("Stroke Rate by Smoking Status")
plt.xlabel("Smoking Status")
plt.ylabel("Stroke Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

## Stroke Rate by Marital Status

```python
married_stroke_rate = pd.DataFrame({
    "marital_status": ["No", "Yes"],
    "stroke_rate": [
        df.loc[df["ever_married_No"] == 1, "stroke"].mean() * 100,
        df.loc[df["ever_married_Yes"] == 1, "stroke"].mean() * 100
    ]
})
married_stroke_rate
```

```python
married_stroke_rate.plot(
    x="marital_status",
    y="stroke_rate",
    kind="bar",
    legend=False
)
plt.title("Stroke Rate by Marital Status")
plt.xlabel("Marital Status")
plt.ylabel("Stroke Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

Marital status acts as a proxy for age. Older age groups have higher stroke rate and are more likely to be married.

## Stroke Rate by Residence Type

```python
residence_stroke_rate = pd.DataFrame({
    "residence_type": ["Rural", "Urban"],
    "stroke_rate": [
        df.loc[df["Residence_type_Rural"] == 1, "stroke"].mean() * 100,
        df.loc[df["Residence_type_Urban"] == 1, "stroke"].mean() * 100
    ]
})
residence_stroke_rate
```

```python
residence_stroke_rate.plot(
    x="residence_type",
    y="stroke_rate",
    kind="bar",
    legend=False
)
plt.title("Stroke Rate by Residence Type")
plt.xlabel("Residence Type")
plt.ylabel("Stroke Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

## Stroke Rate by Work Type

```python
work_stroke_rate = pd.DataFrame({
    "work_type": ["Govt Job", "Never Worked", "Private", "Self-employed", "Children"],
    "stroke_rate": [
        df.loc[df["work_type_Govt_job"] == 1, "stroke"].mean() * 100,
        df.loc[df["work_type_Never_worked"] == 1, "stroke"].mean() * 100,
        df.loc[df["work_type_Private"] == 1, "stroke"].mean() * 100,
        df.loc[df["work_type_Self-employed"] == 1, "stroke"].mean() * 100,
        df.loc[df["work_type_children"] == 1, "stroke"].mean() * 100
    ]
})
work_stroke_rate
```

```python
work_stroke_rate.plot(
    x="work_type",
    y="stroke_rate",
    kind="bar",
    legend=False
)
plt.title("Stroke Rate by Work Type")
plt.xlabel("Work Type")
plt.ylabel("Stroke Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

## Findings Summary

- Stroke cases are much fewer than non-stroke cases.
- Stroke patients are generally older.
- Stroke patients generally have higher glucose levels.
- Hypertension and heart disease show higher stroke rates.
- BMI looks fairly similar between the two groups.
- Male and female stroke rates are quite similar.
- Former smokers show a higher stroke rate than the other smoking groups.
- Rural and urban stroke rates are quite similar.

## Stroke Class Distribution

```python
stroke_distribution = df["stroke"].value_counts(normalize=True) * 100
stroke_distribution
```

Only 5% of patients have stroke. This proves imbalance in data

```python
stroke_distribution.plot(kind="bar")
plt.title("Stroke Class Distribution")
plt.xlabel("Stroke")
plt.ylabel("Percentage (%)")
plt.xticks([0, 1], ["No Stroke", "Stroke"], rotation=0)
plt.show()
```

## Correlation with Stroke

```python
correlation = df.corr()["stroke"].sort_values(ascending=False)
correlation
```

Age has strongest correlation with stroke in dataset.


```python
correlation.drop("stroke").plot(kind="bar", figsize=(10, 5))
plt.title("Correlation of Features with Stroke")
plt.xlabel("Features")
plt.ylabel("Correlation")
plt.xticks(rotation=90)
plt.show()
```

## Key Risk Factors Together

```python
risk_factor_stroke_rate = df.groupby(
    ["hypertension", "heart_disease"]
)["stroke"].mean() * 100
risk_factor_stroke_rate
```

| Hypertension | Heart disease | Stroke rate |
|---|---|---:|
| No | No | 3.39% |
| No | Yes | 16.04% |
| Yes | No | 12.21% |
| Yes | Yes | 20.31% |

Heart disease has stronger association with stroke.

## Correlation Heatmap

```python
plt.figure(figsize=(14, 10))

sns.heatmap(
    df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

Heatmap shows that age, heart disease, avg glucose level and hypertension have strongest positive relationship with stroke.

## Pair Plot of Key Numerical Features

```python
sns.pairplot(
    df[["age", "bmi", "avg_glucose_level", "stroke"]],
    hue="stroke"
)
plt.show()
```

## Key Findings & Insights

- The dataset contains 5,110 records and 22 features after preprocessing.
- The dataset is highly imbalanced, with 95.13% non-stroke cases and 4.87% stroke cases.
- Age shows the strongest positive correlation with stroke among the analyzed features.
- Patients with hypertension have a higher stroke rate (13.25%) than those without hypertension (3.97%).
- Patients with heart disease have a higher stroke rate (17.03%) than those without heart disease (4.18%).
- Patients with both hypertension and heart disease have the highest observed stroke rate (20.31%).
- Average glucose level shows a positive relationship with stroke.
- BMI shows a relatively weak relationship with stroke compared with age, heart disease, glucose level, and hypertension.
- Stroke rates are relatively similar across genders.
- Overall, age, heart disease, hypertension and average glucose level stand out as important factors associated with stroke in this dataset.

## Conclusion

The EDA helped us understand the main patterns in the dataset. Age, heart disease, hypertension, and glucose level showed a noticeable relationship with stroke. The dataset also has much fewer stroke cases than non-stroke cases. These findings can be useful for the next steps of the NeuroShield AI project.
