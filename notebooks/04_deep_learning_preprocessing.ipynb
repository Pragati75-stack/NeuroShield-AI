```python
import pandas as pd
import numpy as np

df = pd.read_csv("../DeepLearning dataset/Original Dataset/multi_source_stroke.csv")

print("Original shape:", df.shape)
df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
missing = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

missing = missing[missing["Missing Count"] > 0]

print(missing)
```

```python
print("Total rows:", len(df))
print("Unique patients:", df["patient_id"].nunique())
print("Duplicate rows:", df.duplicated().sum())
```

```python
target_consistency = df.groupby("patient_id")["stroke_target"].nunique()

print(target_consistency.value_counts())
```

```python
df_patient = df.drop_duplicates(subset="patient_id").copy()

print("Patient-level shape:", df_patient.shape)
```

```python
df_patient = df_patient.drop(columns=["patient_id"])

print(df_patient.shape)
```

```python
print(df_patient["stroke_target"].value_counts())

print("\nPercentage:")
print(
    df_patient["stroke_target"]
    .value_counts(normalize=True) * 100
)
```

```python
clinical_columns = [
    "age",
    "bmi",
    "avg_ap_hi",
    "avg_ap_lo",
    "avg_glucose_level",
    "avg_RestingBP",
    "heart_cholesterol",
    "avg_MaxHR",
    "avg_Oldpeak",
    "avg_height",
    "avg_weight"
]

print(df_patient[clinical_columns].describe().T)
```

```python
for col in clinical_columns:
    print(f"\n{col}")
    print("Zero values:", (df_patient[col] == 0).sum())
```

```python
for col in clinical_columns:
    print(f"\n--- {col} ---")
    print("Minimum:", df_patient[col].min())
    print("Maximum:", df_patient[col].max())
```

```python
# Replace invalid zero values with NaN

invalid_zero_columns = [
    "age",
    "bmi",
    "heart_cholesterol"
]

for col in invalid_zero_columns:
    df_patient.loc[df_patient[col] == 0, col] = np.nan

print("Invalid zero values replaced with NaN.")
```

```python
print("Missing values after clinical validation:")

print(
    df_patient[
        ["age", "bmi", "heart_cholesterol"]
    ].isnull().sum()
)
```

```python
print(
    df_patient[
        df_patient["avg_ap_lo"] > 120
    ][["avg_ap_hi", "avg_ap_lo", "stroke_target"]]
)
```

```python
print(
    "Records with avg_ap_lo > 200:",
    (df_patient["avg_ap_lo"] > 200).sum()
)
```

```python
missing_summary = df_patient.isnull().sum()

print("Missing values in each column:")
print(missing_summary[missing_summary > 0])
```

```python
print("Data types:")
print(df_patient.dtypes)
```

```python
categorical_columns = df_patient.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("Categorical columns:")
print(categorical_columns)
```

```python
# Check missing values in categorical columns

print("Missing values in categorical columns:")
print(df_patient[categorical_columns].isnull().sum())
```

```python
# Check unique values in categorical columns

for col in categorical_columns:
    print(f"\n{col}:")
    print(df_patient[col].value_counts(dropna=False))
```

```python
# Fill missing gender with the most frequent category

df_patient["gender"] = df_patient["gender"].fillna(
    df_patient["gender"].mode()[0]
)

print("Missing gender values:", df_patient["gender"].isnull().sum())
```

```python
# Check all remaining missing values

missing_values = df_patient.isnull().sum()

print("Remaining missing values:")
print(missing_values[missing_values > 0])
```

```python
# Calculate missing percentage for every column

missing_percentage = (
    df_patient.isnull().mean() * 100
).sort_values(ascending=False)

print("Missing percentage:")
print(missing_percentage)
```

```python
# Remove columns with more than 50% missing values

high_missing_columns = missing_percentage[
    missing_percentage > 50
].index.tolist()

print("Columns removed:")
print(high_missing_columns)

df_patient = df_patient.drop(columns=high_missing_columns)

print("\nNew dataset shape:")
print(df_patient.shape)
```

```python
# Check remaining missing values after removing highly incomplete columns

remaining_missing = df_patient.isnull().sum()

print("Remaining missing values:")
print(remaining_missing[remaining_missing > 0])
```

```python
# Check median values of numerical columns with missing data

numerical_missing_columns = [
    "age",
    "bmi",
    "avg_RestingBP",
    "heart_cholesterol",
    "avg_MaxHR",
    "avg_Oldpeak",
    "heart_disease_rate"
]

print("Median values:")
print(df_patient[numerical_missing_columns].median())
```

```python
# Fill missing numerical values using median

for col in numerical_missing_columns:
    df_patient[col] = df_patient[col].fillna(df_patient[col].median())

print("Missing values after median imputation:")
print(df_patient[numerical_missing_columns].isnull().sum())
```

```python
# Check for duplicate records

duplicate_count = df_patient.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Check for invalid clinical values

print("Age below 0:", (df_patient["age"] < 0).sum())
print("BMI below 0:", (df_patient["bmi"] < 0).sum())
print("Resting BP below 0:", (df_patient["avg_RestingBP"] < 0).sum())
print("Cholesterol below 0:", (df_patient["heart_cholesterol"] < 0).sum())
print("Max HR below 0:", (df_patient["avg_MaxHR"] < 0).sum())
print("Oldpeak below 0:", (df_patient["avg_Oldpeak"] < 0).sum())
print("Heart disease rate below 0:", (df_patient["heart_disease_rate"] < 0).sum())
```

```python
# Check target values

print("Stroke target values:")
print(df_patient["stroke_target"].value_counts())
```

```python
# Encode categorical variables using one-hot encoding

categorical_columns = [
    "gender",
    "ever_married",
    "work_type",
    "Residence_type",
    "smoking_status"
]

df_patient = pd.get_dummies(
    df_patient,
    columns=categorical_columns,
    drop_first=True,
    dtype=int
)

print("Dataset shape after encoding:")
print(df_patient.shape)

print("\nData types:")
print(df_patient.dtypes)
```

```python
# Final validation after preprocessing

print("Final dataset shape:")
print(df_patient.shape)

print("\nRemaining missing values:")
print(df_patient.isnull().sum().sum())

print("\nNon-numeric columns:")
print(df_patient.select_dtypes(include=["object"]).columns.tolist())

print("\nStroke target distribution:")
print(df_patient["stroke_target"].value_counts())

print("\nFirst 5 rows:")
display(df_patient.head())
```

```python
# Save the final preprocessed dataset

output_path = "../DeepLearning dataset/Pre-processed Dataset/preprocessed_stroke.csv"

df_patient.to_csv(output_path, index=False)

print("Preprocessed dataset saved successfully!")
print("Location:", output_path)
print("Final shape:", df_patient.shape)
```
